```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("session4_dummy_students_data.csv")

print(df.head())
print(df.info())
print(df.dtypes)
print("Shape:",df.shape)
print("Columns:", df.columns.tolist())
```

```python
#numpy completion

#parameters

def calculate_average(values):
    if len(values)==0:
        return None #Ye nahi krenge to Zero Divison Error aayega
    return sum(values)/len(values)
scores = [77,90,87,45]
result = calculate_average(scores)
print(result)
```

```python
#Column Cleaning: Changes columns names only, not data.
df.columns = (
    df.columns
    .str.lower()
    .str.strip()
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)
print(df.columns.tolist())
```

```python
df["student_id"] = df["student_id"].astype("string")
df["city"] = df["city"].astype("category")
```

```python
#MISSING VALUES

missing_demo = df.copy()

missing_demo.loc[2, "science"] = np.nan
missing_demo.loc[8, "attendance_percent"] = np.nan
missing_demo.loc[14, "city"] = None

print(missing_demo.isna().sum())
```

```python
science_median = missing_demo["science"].median()
attendance_median = missing_demo["attendance_percent"].median()
city_mode = missing_demo["city"].mode(0)

missing_demo["science"] = missing_demo["science"].fillna(science_median)
missing_demo["attendance"] = missing_demo["attendance_percent"].fillna(attendance_median)
missing_demo["city"] = missing_demo["city"].fillna(city_mode)
```

```python
#Create a copy named practice_df. Make the math value in the row 0 missing. Fill the missing math value with math median
# Print the missing Value
practice_df = df.copy()
practice_df.loc[0,"math"] = np.nan
math_median = practice_df["math"].median()
practice_df["math"] = practice_df["math"].fillna(math_median)
print("Filled value for row 0:", practice_df.loc[0, "math"] )
print(math_median)
```

```python
#Averages and some basic Summaries
# Deroving a new column (feature curation) for each student we will calculate their subject averages
df["average_score"] = (
    df[["math","science","english"]]
    .mean(axis=1)
    .round(2)
)
print(df[["name","math","science","english","average_score"]].head())
print("Class average:", df["average_score"].mean())
print("Lowest average:", df["average_score"].min())
print("Highest average:", df["average_score"].max())
print("Students:",df["student_id"].count())

print(df["average_score"].describe()) #Generates a quick statistical summary of a DataFrame
```

```python
#Count mean, min, max
#idxmax() = returns the index label of the maximum value in a Series or along a specified axis of a DataFrame
#loc = a property used to select, filter, and modify data in a DataFrame using label-based indexing
top_student = df.loc[df["average_score"].idxmax(),["name", "average_score"] ]
print(top_student)

low_attendance = df[df["attendance_percent"]<75]
print(low_attendance[["name","attendance_percent","average_score"]])
```

```python
#HomeWork:
#Group the data by:
#same city
#top 25 np.percentile
#Based on these groups, identify some insights
print(df["city"]) #Used this to check all the available city catgories
#Grouping by city
#Finding mean average_score for each city respectivrly to find which cities perform better and which is the best overall
city_performance = df.groupby("city")['average_score'].mean().reset_index()
#reset_index():Resets the index of the DataFrame, and uses the default one instead.
city_performance = city_performance.sort_values(by='average_score', ascending=False)
print('Average Scores Citywise')
print(city_performance)
#Calculating the top 25 percentile
#The top 25 students should be the ones who are 75 percentile achievers
threshold_75 = np.percentile(df['average_score'].dropna(), 75)
#Creating a new dataframe for students who satisfy the threshold
top_25 = df[df['average_score']>= threshold_75]
print('Top 25% Threshold Score:', threshold_75)
print(top_25[['name','city','average_score']])
#Comibining both grouping conditions to Derive Insights
top_student_info = top_25.groupby('city').size().reset_index(name="number_of_top_students")
#Grouped top_25 df with city to know about the smartest kids
top_student_info = top_student_info.sort_values(by='number_of_top_students', ascending = False)

print('City Distribution of Top 25% Students')
print(top_student_info)
```

Based on the calculations, it was observed that majority of top performing students are from Delhi and Noida. It was also observed that city averages of Mathura and Mumbai were way lesser than Delhi and Noida.
