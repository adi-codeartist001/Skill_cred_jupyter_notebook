```python
#Vectors
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
#Study Hours, Attendance, Previous Marks
student_a= np.array([6,85,78])
student_b= np.array([4,75,68])

print("Student A vector", student_a)
print("Student B vector", student_b)
```

```python
#SOTA - State of the art
students = pd.DataFrame({
    "Name":['Arav', 'Bani', 'Chirag', 'Devi'],
    "study_hours":[85,70,90,60],
    "previous_marks":[78,85,88,35],
    "final_marks":[80,90,95,88]
})
students
```

```python
#Plotting a scatter plot of students
plt.figure(figsize=(8,6))
plt.scatter(students["study_hours"], students["previous_marks"], color="blue")
for i,row in students.iterrows():
    plt.annotate(row["Name"], (row["study_hours"], row["previous_marks"]))
plt.title("Scatter Plot of Study Hours vs Marks")
plt.xlabel("Name")
plt.ylabel("Marks")
plt.show()
```

```python
#Finding Euclidean distance
point_a = np.array([6,85])
point_b = np.array([3, 70])

distance = np.sqrt((point_a[0] - point_b[0]) ** 2 + (point_a[1] - point_b[1]) ** 2)

print ("distance", distance)
```
