```python
import pandas as pd
```

```python
data = {
    "student_id": [101, 102, 103, 104, 105, 106],
    "attendance_percent": [92, 67, 81, 45, 74, 88],
    "assignment_score": [18, 12, 15, 8, 14, 19],
    "quiz_score": [72, 48, 65, 30, 55, 80],
    "lab_completed": [True, False, True, False, True, True]
}
df = pd.DataFrame(data)
print(df)
```

```python
#Feature Extraction
df["total_score"] = df["assignment_score"] + df["quiz_score"]
df
```

```python
#Eligibility Rule for scholarship
#If Attendance>=75 and total_score>=70 and if labs are completed
df["eligible"] = ((df["attendance_percent" ]>=75) &
                 (df['total_score'] >= 70)&
                 (df['lab_completed'] == True) )
df
```

```python
#Filter
eligible_students = df[df["eligible"] == True]
eligible_students
```

```python
summary = {
    "average_attendance": df["attendance_percent"].mean(),
    "average_total_score": df["total_score"].mean(),
    "eligible_count": df["eligible"].sum(),
    "total_students": len(df)
}
summary
```

```python
#Assertions For Attendance Percentage, Assignment Score, and Quiz Score
assert df["attendance_percent"].between(0,100).all(),"Attendance percentage should be between 0 and 100"
assert df['assignment_score'].between(0,20).all(), "Assignment score should be between 0 and 20"
```
